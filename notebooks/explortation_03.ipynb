```python
from src.preprocess.parquet_preprocessor import ParquetPreprocessor

preprocessor = ParquetPreprocessor()
preprocessor.csv_to_parquet("data/raw/alarm_history_dump.csv", "data/raw/alarm_history_dump.parquet")
```

```python
import polars as pl
from IPython.display import display

df = pl.scan_parquet("data/raw/alarm_history_dump.parquet")

display(df.head().collect())
```

```python
from src.preprocess.sequence_preprocessor2 import SequencePreprocessor2

preprocessor = SequencePreprocessor2()
query = preprocessor.select_features(df)
query = preprocessor.clean_data(query)
```

```python
alarm_by_node = (
    query.group_by("Node Name")
    .agg(
        pl.col("Alarm ID").n_unique().alias("Quantidade de Alarmes")
    )
    .sort("Quantidade de Alarmes", descending=True)
)

display(
    alarm_by_node.collect()
    .to_pandas()
    .style.set_table_attributes("style='display:inline-block; max-height:300px; overflow-y:auto;'")
)
```
