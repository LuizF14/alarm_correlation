```python
from notebooks.utils.view_network import plot_graph
import igraph as ig

edges_df = pl.read_parquet('data/models/simple_time_correlation_historypreprocessor_edges/part_65.parquet')

all_nodes = list(set(edges_df["src"].to_list() + edges_df["dst"].to_list()))
node_index = {n: i for i, n in enumerate(all_nodes)}

g = ig.Graph(
    n=len(all_nodes),
    edges=[(node_index[s], node_index[d]) for s, d in zip(edges_df["src"].to_list(), edges_df["dst"].to_list())],
    directed=True,
)
g.vs["name"] = all_nodes

plot_graph(g, "part_65")
```
