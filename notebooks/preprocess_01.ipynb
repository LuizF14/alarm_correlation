```python
import pandas as pd

df = pd.read_csv('data/raw/active_alarms_prod.csv')
```

```python
from IPython.display import display

def scroll_display(df):
    display(df.style.set_table_attributes(
        "style='display:inline-block; max-height:300px; overflow-y:auto; border: 1px solid #ccc;'"
    ))
```

```python
from src.preprocess.sequence_preprocessor import SequencePreprocessor

preprocessor = SequencePreprocessor()

uncleaned_df = preprocessor.select_features(df)
cleaned_df = preprocessor.clean_data(uncleaned_df)
```

```python
alerts_by_node_uncleaned = (uncleaned_df.groupby('Node Name')['Alert ID']
                  .nunique()
                  .sort_values(ascending=False)
                  .rename_axis('Node Name')
                  .to_frame(name='Quantidade de Alertas'))

alerts_by_node_cleaned = (cleaned_df.groupby('Node Name')['Alert ID']
                  .nunique()
                  .rename_axis('Node Name')
                  .to_frame(name='Quantidade de Alertas'))

comparison_df = pd.merge(
    alerts_by_node_uncleaned, 
    alerts_by_node_cleaned, 
    on='Node Name', 
    how='outer', 
    suffixes=(' Sujo', ' Limpo')
)

comparison_df = comparison_df.sort_values(by='Quantidade de Alertas Sujo', ascending=False)
scroll_display(comparison_df)
```

```python
node_alerts = cleaned_df[cleaned_df['Node Name'] == 'SPCTP-IFOHPP-STR04'].sort_values('First Occurrence')

scroll_display(node_alerts)
```
