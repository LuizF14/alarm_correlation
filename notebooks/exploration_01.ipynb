```python
import pandas as pd
from IPython.display import display
```

```python
df = pd.read_csv('data/raw/active_alarms_prod.csv')
df.head()
```

```python
df.info()
```

```python
alerts_by_node = (df.groupby('acal_cd_node_id')['acal_cd_alarm_id']
                  .nunique()
                  .sort_values(ascending=False)
                  .rename_axis('Node ID') # Renomeia o índice
                  .to_frame(name='Quantidade de Alertas'))
display(alerts_by_node.style.set_table_attributes("style='display:inline-block; max-height:300px; overflow-y:auto;'"))
```

```python
alerts_by_type = (df.groupby('acal_tx_alert_type')['acal_cd_alarm_id']
                  .count()
                  .sort_values(ascending=False)
                  .rename_axis('Tipo de Alerta')
                  .to_frame(name='Total de Ocorrências'))

display(alerts_by_type.style.set_table_attributes("style='display:inline-block; max-height:300px; overflow-y:auto;'"))
```

```python
def scroll_display(df):
    display(df.style.set_table_attributes(
        "style='display:inline-block; max-height:300px; overflow-y:auto; border: 1px solid #ccc;'"
    ))


alertas_por_node = df.groupby('acal_tx_node')['acal_tx_alert_type'].nunique().reset_index()
alertas_por_node.columns = ['Node Name', 'Qtd Alert Types Únicos']

alertas_por_node = alertas_por_node.sort_values("Qtd Alert Types Únicos", ascending=False).reset_index(drop=True)

scroll_display(alertas_por_node)
```

```python
node_alerts = df[df['acal_tx_node'] == 'SPCTP-IFOHPP-STR04'].sort_values('acal_dt_first_occurrence')

view_cols = [
    'acal_dt_first_occurrence', 
    'acal_tx_alert_type', 
    'acal_nm_severity', 
    'acal_tx_description'
]

scroll_display(node_alerts[view_cols])
```
