```python
import pandas as pd
from IPython.display import display
```

```python
df = pd.read_csv('data/raw/active_alarms_prod.csv')
df.head()
```

```python
df.info()
```

```python
alerts_by_node = (df.groupby('acal_tx_node')['acal_cd_alarm_id']
                  .nunique()
                  .sort_values(ascending=False)
                  .rename_axis('Nome do Node') # Renomeia o índice
                  .to_frame(name='Quantidade de Alertas'))
display(alerts_by_node.style.set_table_attributes("style='display:inline-block; max-height:300px; overflow-y:auto;'"))
```

```python
alerts_by_type = (df.groupby('acal_tx_alert_type')['acal_cd_alarm_id']
                  .count()
                  .sort_values(ascending=False)
                  .rename_axis('Tipo de Alerta')
                  .to_frame(name='Total de Ocorrências'))

display(alerts_by_type.style.set_table_attributes("style='display:inline-block; max-height:300px; overflow-y:auto;'"))
```
